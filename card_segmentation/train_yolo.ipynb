```python
result = prediction[0].plot()
result = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
plt.imshow(result)
```

```python
corners = prediction[0].obb.xyxyxyxy[0].cpu().numpy()
corners
```

```python
center = np.mean(corners, axis=0)
center
```

```python
expanded_corners = (corners - center) * 1.05 + center
expanded_corners
```

```python
def extract_card(image, corners):
    corners = np.int32(corners)
    ordered_corners = utimg.order_points(corners)
    return utimg.four_point_transform(image, ordered_corners)
```

```python
warped = extract_card(prediction[0].orig_img, expanded_corners)
warped = cv2.cvtColor(warped, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(warped)
```

```python
im = cv2.bitwise_not(warped)
plt.imshow(im)
```

```python
saturation = cv2.cvtColor(im, cv2.COLOR_BGR2HSV)[..., 2]
saturation = cv2.medianBlur(saturation, 15)
plt.imshow(saturation)
```

```python
edges = cv2.Canny(saturation, 75, 200, L2gradient = True)
plt.imshow(edges)
```

```python
lines = cv2.HoughLines(edges, 1, np.pi/180, 100)
lines = lines[:, 0, :]
lines = [(line[0], line[1]) for line in lines]
line_num = len(lines)
new_lines = []
new_lines.append(lines[0])
for i in range(1, line_num):
    valid = True
    for line in new_lines:
        if abs(line[0] - lines[i][0]) < 100 and abs(line[1] - lines[i][1]) < 0.3:
            valid = False
            break
    if valid:
        new_lines.append(lines[i])
        
lines = new_lines
lines
```

```python
tmp = im.copy()
```

```python
for line in lines:
    rho,theta = line
    a = np.cos(theta)
    b = np.sin(theta)
    x0 = a*rho
    y0 = b*rho
    x1 = int(x0 + 1000*(-b))
    y1 = int(y0 + 1000*(a))
    x2 = int(x0 - 1000*(-b))
    y2 = int(y0 - 1000*(a))

    cv2.line(tmp, (x1, y1), (x2, y2), (0,0,255), 2)

plt.imshow(tmp)
```

```python
intersection = docdetect.find_intersections(lines, im)
intersection
```

```python
quadrilaterals = docdetect.find_quadrilaterals(intersection)
quadrilaterals
```

```python
frame = docdetect.draw(quadrilaterals, im)
plt.imshow(frame)
```

```python
card = extract_card(warped, [corner['coords'] for corner in intersection])
plt.imshow(card)
```
