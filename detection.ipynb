```python
import matplotlib.pyplot as plt
from classification import load_templates, load_image
from card_detection import detect_card
```

```python
template_dir = "templates"
templates = load_templates(template_dir)
fig, axes = plt.subplots(2, 5, figsize=(20,5))
for i, template in enumerate(templates):
    axes[i // 5, i % 5].imshow(template)
plt.show()
```

```python
image_dir = "image27.png"
image = load_image(image_dir, (512, 1024))
plt.imshow(image)
```

```python
card, label = detect_card(image, templates)
if card is not None:
    plt.imshow(card)
```

```python
print("The label of the card is:", label)
plt.imshow(templates[label])
```
